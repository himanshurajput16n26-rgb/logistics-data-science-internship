# Week 2 — Logistics Data Cleaning & Preprocessing

Reference: DataCo SMART SUPPLY CHAIN FOR BIG DATA ANALYSIS.

```python
import pandas as pd
from pathlib import Path

path = Path('../data/raw/DataCoSupplyChainDataset.csv')
df = pd.read_csv(path, encoding='latin-1')
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df.shape
```

## Missing Values and Duplicates

```python
missing = df.isna().sum().to_frame('missing_count')
missing['missing_pct'] = missing['missing_count'] / len(df) * 100
display(missing.sort_values('missing_pct', ascending=False).head(15))
print('Duplicates:', df.duplicated().sum())
```

## Date Conversion

```python
for col in ['order_date_(dateorders)', 'shipping_date_(dateorders)']:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')
df.head()
```

## IQR Outlier Detection

```python
if 'sales' in df.columns:
    q1 = df['sales'].quantile(0.25)
    q3 = df['sales'].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    df['sales_outlier_flag'] = ((df['sales'] < lower) | (df['sales'] > upper)).astype(int)
    print('Potential outliers:', df['sales_outlier_flag'].sum())
```
